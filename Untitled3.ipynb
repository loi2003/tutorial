```python


file_path = PATH + 'forest.jpg'

# ================== 5. LOAD IMAGE ==================
Img = cv2.imread(file_path)



    # ================== 6. CONVERT BGR -> RGB ==================
Img_color = cv2.cvtColor(Img, cv2.COLOR_BGR2RGB)

    # ================== 7. CONVERT RGB -> GRAYSCALE ==================
Img_gray = cv2.cvtColor(Img_color, cv2.COLOR_RGB2GRAY)

    # ================== 8. THRESHOLD (BINARY) ==================
_, Img_bw = cv2.threshold(Img_gray, 127, 255, cv2.THRESH_BINARY)
```

```python
from cv2.detail import leaveBiggestComponent
#Ex1_3: Sampling and Quantization
Img = cv2.imread(PATH + 'rose.jpg')
#convert from integer to double format (from 0->255 to 0.0000->255.0000)
Img = Img.astype(np.float64)

#quantize and display image in different number of bits (from 1 to 8 bits)
plt.figure(figsize = (20, 10))
for num_bits in range(1,9):
  #number of gray levels
  num_levels = 2 ** num_bits
  #level gap for each level
  level_gap = 256 / num_levels

  #quantize image
  quantized_Img = np.uint8(np.ceil(Img / level_gap) * level_gap - 1)
  quantized_Img = np.clip(quantized_Img, 0, 255)

  #display image
  plt.subplot(2, 4, num_bits), plt.imshow(quantized_Img)
  title = f"{num_bits}-bit" if 1==num_bits else f"{num_bits}-bits"
  plt.title(title, fontsize=FS), plt.axis('off')

plt.tight_layout()
plt.savefig(PATH + 'Ex1_3.jpg', dpi=400)
plt.show()
```
